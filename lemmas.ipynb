```python
import json
import spacy
from spacy.tokens import DocBin
from collections import defaultdict


# Load the LatinCy model
nlp = spacy.load("la_core_web_lg")
```

```python
# Load the original corpus
with open("./Corpus/latin_tragedies_corpus.json", "r", encoding="utf-8") as f:
    raw_corpus = json.load(f)

doc_bin = DocBin()
titles = []

# Serialize tragedies to disk
for entry in raw_corpus:
    doc = nlp(entry["text"])
    doc_bin.add(doc)
    titles.append(entry["title"])

# Save the DocBin to disk
doc_bin.to_disk("./Corpus/latin_tragedies_corpus.spacy")
```

```python
# Target lemmas
target_lemmas = {"fata", "fatum"}

# Dictionary to hold lemma occurrences
lemma_to_words = defaultdict(set)

# Load the DocBin from disk
doc_bin = DocBin().from_disk("./Corpus/latin_tragedies_corpus.spacy")

docs = list(doc_bin.get_docs(nlp.vocab))

# For each lemma, collect all token occurrences with detailed info            
for title, doc in zip(titles, docs):
    for token in doc:
        if token.lemma_ in target_lemmas:
            lemma_to_words[token.lemma_].add((title, token.text, token.sent, token.morph, token.pos_))

# Display results
for lemma, items in lemma_to_words.items():
    print(f"\n{lemma}:")
    for play, word, sentence, morph, pos in sorted(
        items, key=lambda x: (x[0], x[1])
        ):
        print(f"  {play}: {word}")
        print(f"    Line: {sentence}")
        print(f"    Morph: {morph}")
        print(f"    POS: {pos}")
        print()
```
